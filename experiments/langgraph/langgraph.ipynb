```python
from langgraph_sdk import get_client
from google.oauth2 import service_account
from google.auth.transport.requests import Request
```

```python
# Fetch token
credentials = service_account.IDTokenCredentials.from_service_account_file(
            "C:\\code\\chillpill\\api\\chillpill-api_service_account.json",
            target_audience="http://127.0.0.1:2024",
        )

# Refresh credentials to get the token
credentials.refresh(Request())
token = credentials.token
```

```python
print(token)
```

```python
client = get_client(
    url="http://127.0.0.1:2024",
    headers={"Authorization": f"Bearer {token}"}
)
```

```python
thread = await client.threads.create()

# Create a stateless run for the client
# Pass input_idea as the initial state
input_state = {
    "input_idea": "Story of a Uber Driver, struggling to find peace in a chaotic city.",
    "db_metadata":{"project_id": "681796fabbb5d812dc05649e"}
}

graph_name = "kahaaniai"

run = await client.runs.create(thread["thread_id"], graph_name, input=input_state)

print(f"Run ID: {run['run_id']}")
print(f"Status: {run['status']}")
```

```python
# Fetching the run result
result = await client.threads.get_state(thread["thread_id"])

print(f"Run Result: {result}")
```
